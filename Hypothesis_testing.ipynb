```python
import matplotlib.pyplot as plt
import scipy.stats as stat
import numpy as np
```

```python
n_patient = 1000
p_success = 0.9
dist_binomial = stat.binom(n_patient,p_success)
x_values = np.arange(n_patient+1,step=5)
y_values = dist_binomial.pmf(x_values)
plt.stem(x_values,y_values)

```
