```python
import pandas as pd
import numpy as np

data = pd.read_excel("diabetes_dataset_120_cases.xlsx")
data
```

```python
df = data
df['smoking_history'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()

df['smoking_history'] = label_encoder.fit_transform(df['smoking_history'])
```

```python
df
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
pac1 = PCA(n_components=2)
pca2 = PCA(n_components=0.80)

pca_data = pac1.fit_transform(scaled_data)
pca_data2 = pca2.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_data , columns=['PC1','PC2'])
```

```python
print("Explained variance by each principal componenet:" , pac1.explained_variance_ratio_)

print(pca_df)
```
